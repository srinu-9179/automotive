```python
from google.colab.output import eval_js
from IPython.display import display, Javascript
import cv2
import numpy as np
from PIL import Image
import base64, io

def capture_frame():
    js = Javascript('''
      async function takePhoto() {
        const stream = await navigator.mediaDevices.getUserMedia({video: true});
        const video = document.createElement('video');
        video.srcObject = stream;
        await video.play();

        // Capture one frame
        const canvas = document.createElement('canvas');
        canvas.width = video.videoWidth;
        canvas.height = video.videoHeight;
        canvas.getContext('2d').drawImage(video, 0, 0);

        // Stop webcam
        stream.getTracks().forEach(track => track.stop());

        return canvas.toDataURL('image/jpeg');
      }
      takePhoto();
    ''')

    data = eval_js(js)
    data = data.split(',')[1]
    img_bytes = base64.b64decode(data)
    img = Image.open(io.BytesIO(img_bytes))
    return img
```

```python

from google.colab import output
import cv2, numpy as np, PIL.Image, io, base64
from IPython.display import display, Javascript
import time

captured_image = None

def js_capture():
    display(Javascript('''
        async function takePhoto() {

            // Start webcam
            const stream = await navigator.mediaDevices.getUserMedia({video: true});
            const video = document.createElement('video');
            video.style.display = 'block';
            video.srcObject = stream;

            // Display video so Colab keeps it alive
            document.body.appendChild(video);

            await video.play();

            // KEEP WEBCAM ON FOR 1 SECOND BEFORE CAPTURE
            await new Promise(resolve => setTimeout(resolve, 1000));

            // Capture frame
            const canvas = document.createElement('canvas');
            canvas.width = video.videoWidth;
            canvas.height = video.videoHeight;
            canvas.getContext('2d').drawImage(video, 0, 0);

            // Stop webcam
            stream.getTracks().forEach(track => track.stop());

            // Remove video preview
            video.remove();

            // Send data
            const data = canvas.toDataURL('image/jpeg', 0.8);
            google.colab.kernel.invokeFunction('notebook.capture', [data], {});
        }
        takePhoto();
    '''))
```

```python
def callback(data):
    global captured_image
    img_str = data.split(",")[1]
    captured_image = PIL.Image.open(io.BytesIO(base64.b64decode(img_str)))

output.register_callback('notebook.capture', callback)
```

```python
!pip install ultralytics
from ultralytics import YOLO

model = YOLO("/content/runs/detect/train/weights/best.pt")  # update path if needed
```

```python
from google.colab.patches import cv2_imshow

print("Continuous Webcam Detection Started")
print("Press STOP ON THE CELL to exit.\n")

while True:
    captured_image = None

    js_capture()  # start webcam → wait → capture frame

    # Wait for the callback to finish
    timeout = time.time() + 5  # 5-second timeout
    while captured_image is None and time.time() < timeout:
        time.sleep(0.1)

    if captured_image is None:
        print("⚠ No frame captured. Retrying...")
        continue

    # Convert to numpy array
    frame_np = np.array(captured_image)

    # Run YOLO detection
    results = model.predict(frame_np, conf=0.5)
    annotated = results[0].plot()

    # Show output
    cv2_imshow(annotated)

    # Wait before next frame
    time.sleep(1)
```

```python
from IPython.display import display, Javascript, Image
import cv2, base64, numpy as np
from google.colab.output import eval_js

def start_webcam():
    display(Javascript('''
        var video;
        var canvas;
        var ctx;

        navigator.mediaDevices.getUserMedia({ video: true })
        .then(function(stream) {
          video = document.createElement('video');
          document.body.appendChild(video);
          video.srcObject = stream;
          video.stop();

          canvas = document.createElement('canvas');
          ctx = canvas.getContext('2d');
        });
    '''))
```

```python
def get_frame():
    js = Javascript('''
        async function captureFrame() {
            const canvas = document.querySelector("canvas");
            const video = document.querySelector("video");
            canvas.width = video.videoWidth;
            canvas.height = video.videoHeight;
            canvas.getContext('2d').drawImage(video, 0, 0);
            return canvas.toDataURL('image/jpeg', 0.8);
        }
        return captureFrame();
    ''')
    data = eval_js(js)
    data = data.split(',')[1]
    img = np.frombuffer(base64.b64decode(data), dtype=np.uint8)
    return cv2.imdecode(img, cv2.IMREAD_COLOR)
```

```python
from google.colab.patches import cv2_imshow
import time

print("Starting helmet detection...")

while True:
    frame = get_frame()

    if frame is None:
        print("⚠ No frame captured, retrying...")
        time.sleep(0.5)
        continue

    results = model.predict(frame, conf=0.5)
    annotated = results[0].plot()

    cv2_imshow(annotated)

    time.sleep(1)   # capture every 1 second
```
